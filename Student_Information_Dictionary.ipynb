```python

# Student Information Dictionary
# Veda Technology - Day 7 Task

students = {
    101: {"name": "Arun", "age": 21, "course": "MCA", "marks": 85},
    102: {"name": "Priya", "age": 22, "course": "MCA", "marks": 90},
    103: {"name": "Kumar", "age": 21, "course": "MCA", "marks": 78}
}

# Display all student records
def display_students():
    print("Student Records:")
    for student_id, details in students.items():
        print(student_id, details)

# Search for a student
def search_student(student_id):
    if student_id in students:
        print("Student Found:", students[student_id])
    else:
        print("Student ID not found.")

# Add a new student
def add_student(student_id, name, age, course, marks):
    if student_id not in students:
        students[student_id] = {
            "name": name,
            "age": age,
            "course": course,
            "marks": marks
        }
        print("Student added successfully.")
    else:
        print("Student ID already exists.")

# Update an existing student
def update_student(student_id, marks):
    if student_id in students:
        students[student_id]["marks"] = marks
        print("Student record updated successfully.")
    else:
        print("Student ID not found.")

# Perform operations
display_students()

print("\n--- Search Operation ---")
search_student(101)

print("\n--- Add Operation ---")
add_student(104, "Divya", 22, "MCA", 88)

print("\n--- Update Operation ---")
update_student(102, 95)

print("\n--- Final Student Records ---")
display_students()
```
